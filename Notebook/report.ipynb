Här visar diagrammet vilka kategorier som genererar högst intäkter. Där ser vi att elektronik, sport och kläder genererar mest

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("../Data/ecommerce_sales.csv")

yt = [1000,2000,3000,4000,5000,6000,7000,8000,9000,10000]

plt.bar(x=data["category"], height=data["revenue"])
plt.xlabel("Kategori")
plt.ylabel("SEK")
plt.yticks(yt)
plt.title("Intäkter")
plt.show()
```

Här visar diagrammet att Stockholm, Göteborg och Västerås genererar högst intäkter

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("../Data/ecommerce_sales.csv")


plt.bar(x=data["city"], height=data["revenue"])
plt.xlabel("Stad")
plt.ylabel("SEK")
plt.title("Intäkter")
plt.show()
```

Har gjort en funktion för att räkna ut den totala intäkten och antal enheter. 

```python
def total(rev, unit):
    return rev.astype(int).sum(), unit.sum()

result = total(data["revenue"], data["units"])
print(result)
```

Här har jag sammanställt de totala intäkterna per kategori och stad. Har även fått ut genomsnittliga ordervärdet

```python
import pandas as pd

data = pd.read_csv("../Data/ecommerce_sales.csv")

cat_rev = data.groupby("category")["revenue"].sum()

city_rev = data.groupby(data["city"])["revenue"].sum()

aov_data = data["revenue"].sum() / len(data)

print(f"Intäkter per kategori\n{cat_rev}\n")
print(f"Intäkter per stad\n{city_rev}\n")
print(f"Genomsnittligt ordervärde: {aov_data}")
```
